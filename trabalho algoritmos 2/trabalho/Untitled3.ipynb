```python
# Tabela de substituição de vogais (inclui maiúsculas e minúsculas com acentos)
substituicoes = {
    'a': '!', 'A': '!', 'á': '!', 'à': '!', 'ã': '!', 'â': '!', 'Á': '!', 'À': '!', 'Ã': '!', 'Â': '!',
    'e': '@', 'E': '@', 'é': '@', 'ê': '@', 'É': '@', 'Ê': '@',
    'i': '#', 'I': '#', 'í': '#', 'Í': '#',
    'o': '$', 'O': '$', 'ó': '$', 'ô': '$', 'õ': '$', 'Ó': '$', 'Ô': '$', 'Õ': '$',
    'u': '%', 'U': '%', 'ú': '%', 'Ú': '%'
}

# Solicita o nome do arquivo de saída ao usuário
nome_saida = input("Digite o nome do arquivo de saída (ex: musica_codificada.txt): ")

try:
    # Leitura do arquivo original
    with open("musica.txt", "r", encoding="utf-8") as arquivo_entrada:
        conteudo = arquivo_entrada.read()

    # Processamento e substituição dos caracteres
    texto_codificado = ""
    for caractere in conteudo:
        if caractere in substituicoes:
            texto_codificado += substituicoes[caractere]
        else:
            texto_codificado += caractere

    # Gravação do novo arquivo codificado
    with open(nome_saida, "w", encoding="utf-8") as arquivo_saida:
        arquivo_saida.write(texto_codificado)

    print("Conversão concluída com sucesso!")

except FileNotFoundError:
    print("Erro: O arquivo txt não foi encontrado no mesmo diretório do script.")
```
